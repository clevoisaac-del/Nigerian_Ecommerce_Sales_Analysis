Nigerian E-commerce Sales Analysis

Project Overview

This project analyzes a Nigerian e-commerce sales dataset to understand sales performance, order trends, product performance, branch performance, and order fulfillment outcomes.

The analysis was performed using Python, Pandas, Matplotlib, and Excel.

Project Objectives

- Analyze overall sales performance
- Identify monthly sales trends
- Compare branch performance
- Analyze delivered, cancelled, invalid, and failed orders
- Identify top-performing products
- Examine sales per unit
- Create visualizations to communicate business insights
- Provide data-driven recommendations for improving sales and order fulfillment

Key Business Questions

1. How much revenue was generated?
2. Which branches generated the most sales?
3. How did sales performance change over time?
4. Which products performed best?
5. What proportion of orders were cancelled or delivered?
6. What business areas require improvement?

Tools Used

- Python
- Pandas
- Matplotlib
- Jupyter Notebook
- Microsoft Excel
- Data Analysis
- Data Visualization

Project Summary and Key Metrics

This project analyzes Nigerian e-commerce sales data from February to May 2021 to evaluate revenue performance, product contributions, branch performance, and order outcomes.

Key Metrics

- Total Sales: Approximately ₦1.78 billion
- Total Orders: 2,406
- Total Units Sold: 222,719
- Average Order Value: Approximately ₦741,082.83
- Top 10 Products' Share of Sales: 58.43%

Business Focus

The analysis investigates sales trends, differences in branch performance, product sales concentration, and order cancellations to identify opportunities for better inventory planning, sales management, and order fulfillment.

```python
def get_final_status(statuses):
    statuses = set(statuses)

    if "Delivered" in statuses:
        return "Delivered"
    elif "Cancelled" in statuses:
        return "Cancelled"
    elif "Invalid order" in statuses:
        return "Invalid order"
    elif "Failed delivery" in statuses:
        return "Failed delivery"
    elif "Packed" in statuses:
        return "Packed"
```

```python
def choose_tool(request):
    request = request.lower()

    if "total sales" in request:
        return "calculate_total_sales"

    elif "top products" in request or "top 5 products" in request:
        return "find_top_products"

    elif "monthly sales" in request or "sales by month" in request:
        return "get_monthly_sales"

    else:
        return "No suitable tool found"
```
